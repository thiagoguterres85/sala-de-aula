```python

class Plataforma():
    nome: str
    fabricante: str 

    def __init__(self, nome:str, fabricante:str):
        self.nome=nome
        self.fabricante=fabricante

    def __str__(self):
        return f"Plataforma(nome[{self.nome}, fabricante={self.fabricante}])"

    def __repr__(self):
        return f"Plataforma {self.nome}"
```

```python
for plataforma in plataformas_para_inserir:
    try:
        cursor.execute(
            "INSERT INTO plataforma (nome, fabricante) VALUES (%s,%s);",(plataforma.nome, plataforma.fabricante)
        )
        conexao.commit()
    except pg.DatabaseError as e:
        print(e)
        conexao.rollback()

    
```
