```python

class Plataforma():
    nome: str
    fabricante: str 

    def __init__(self, nome:str, fabricante:str):
        self.nome=nome
        self.fabricante=fabricante

    def __str__(self):
        return f"Plataforma(nome[{self.nome}, fabricante={self.fabricante}])"

    def __repr__(self):
        return f"Plataforma {self.nome}"
```

```python
plataformas_para_inserir: list[Plataforma] = [ 
    Plataforma("PlayStation 3", "Sony"),
    Plataforma("PlayStation 2", "Sony"),
    Plataforma("PlayStation 1", "Sony"),
    Plataforma("Xbox Original", "Microsoft"),
    Plataforma("Xbox 360", "Microsoft"),
    Plataforma("XBOX Series X", "Microsoft"),]

print(plataformas_para_inserir)
```

```python
class Cliente():
    id_cliente:int
    nome: str
    telofone: str 
    email: str

    def __init__(self, nome:str, telefone:str, email:str):
        self.nome=nome
        self.telefone=telefone
        self.email=email
    def __str__(self):
        return f"Cliente(nome[{self.nome} telefone={self.telefone}, email={self.email}])"

    def __repr__(self):
        return f"Cliente {self.nome}"
```

```python
class Quadra():
    id_quadra:int           
    preco: float 
    coberta: bool  
    
    def __init__(self, preco:float, coberta:bool):
        self.preco=preco
        self.coberta=coberta
    def __str__(self):
        return f"Quadra (preco[{self.preco}, coberta={self.coberta}])"

    def __repr__(self):
        return f"Quadra {self.nome}"
```

```python
class Partida():
    id_partida: int
    nome_artilheiro: str
    total_gols: int

    def __init__(self, nome_artilheiro:str, total_gols:int):
        self.nome_artilheiro=nome_artilheiro
        self.total_gols=total_gols

    def __str__(self):
        return f"Partida( nome_artilheiro[{self.nome_artilheiro}, total_gols={self.total_gols}])"

    def __repr__(self):
        return f"Partida {self.nome}"
```

```python

class Agendamento():
    id_agendamento: int
    id_quadra: int
    tempo_termino_max: int
    hora: datetime  
    id_partida: int


    def __init__(self, tempo_termino_max: int, hora:datetime):
        self.tempo_termino_max=tempo_termino_max
        self.hora=hora

    def __str__(self):
        return f"Agendamento( tempo_termino_max[{self.tempo_termino_max}, hora={self.hora}])"
    
    def __repr__(self):
        return f"Agendamento {self.nome}"
```
